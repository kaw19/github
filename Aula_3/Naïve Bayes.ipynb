## 1.1 O que é o Naïve Bayes?
O **Naïve Bayes** é um classificador simples, porém poderoso, baseado em um modelo probabilístico derivado do Teorema de Bayes. Basicamente, ele determina a probabilidade de uma instância pertencer a uma classe com base nas probabilidades dos valores de cada característica. 

O termo "*naïve*" (ingênuo) deriva do fato de que ele assume que cada característica é independente das demais; isto é, o valor de uma característica não tem relação com o valor de outra. Apesar de ser uma simplificação forte, é exatamente essa "ingenuidade" que torna o algoritmo extremamente rápido e eficiente em aplicações práticas.

## 1.3 Tipos de Classificação
Na classificação, o objetivo é aprender uma regra geral que mapeia as **observações** (características) para categorias alvo (**rótulos** ou **classes**). Existem três tipos principais:

* <span style="color:#e50914; font-weight:bold;">Classificação Binária:</span> Classifica observações em uma de duas classes possíveis. Exemplo: O filme é "Recomendado" ou "Não Recomendado"? Um e-mail é "Spam" ou "Não Spam"?
* <span style="color:#e50914; font-weight:bold;">Classificação Multiclasse:</span> Permite mais de duas classes possíveis. Exemplo: Reconhecimento de dígitos manuscritos em CEPs (dígitos de 0 a 9).
* <span style="color:#e50914; font-weight:bold;">Classificação Multi-rótulo:</span> Uma mesma amostra pode pertencer a várias categorias simultaneamente. Exemplo: Um filme como *Vingadores* pode ser classificado ao mesmo tempo como Ação, Aventura e Ficção Científica.


# **Capítulo 2: A Matemática da Previsão (Desvendando o Teorema de Bayes)**

Antes de programar, precisamos entender a "mágica" matemática por trás do algoritmo. O classificador Naïve Bayes calcula a distribuição de probabilidade sobre todas as classes possíveis e escolhe aquela com a maior chance de estar correta.

## 2.1 O Teorema de Bayes
O Teorema de Bayes calcula a probabilidade de um evento $A$ ocorrer, dado que um evento $B$ já é verdadeiro. A fórmula clássica é:

$$P(A | B) = \frac{P(B | A) P(A)}{P(B)}$$

Para o nosso algoritmo, traduzimos essa fórmula em quatro componentes fundamentais:

1. **Prior (Probabilidade a priori) - $P(y_k)$:** Como as classes estão distribuídas antes de olharmos os dados. Ex: Sem saber nada sobre um usuário, qual a chance geral de ele gostar de um filme de terror?
2. **Posterior (Probabilidade a posteriori) - $P(y_k|x)$:** É o nosso objetivo. A probabilidade de o usuário gostar do filme *depois* de analisarmos as características dele.
3. **Verossimilhança (Likelihood) - $P(x|y_k)$:** A probabilidade de encontrarmos essas características específicas em alguém que já sabemos que gosta do filme.
4. **Evidência - $P(x)$:** A distribuição geral das características (funciona apenas como uma constante de normalização).

## 2.2 Exemplo Prático: Por que não confiar apenas na intuição?
Para entender o poder do Teorema de Bayes na vida real, vamos a um cenário de diagnóstico médico (rastreamento de câncer) em uma população de 10.000 pessoas:

| Resultado | Com Câncer | Sem Câncer | Total |
| :--- | :--- | :--- | :--- |
| **Teste Positivo** | 80 | 900 | 980 |
| **Teste Negativo** | 20 | 9000 | 9020 |
| **Total** | 100 | 9900 | 10000 |

* O teste acerta 80 dos 100 casos reais de câncer.
* O teste dá um "falso alarme" em 900 pessoas saudáveis.

**A pergunta é:** Se você fizer o teste e o resultado der **POSITIVO**, qual é a probabilidade real de você ter câncer? A intuição humana costuma chutar valores muito altos (como 80%). Mas aplicando o Teorema de Bayes:

$$P(\text{Câncer} | \text{Positivo}) = \frac{P(\text{Positivo} | \text{Câncer}) \times P(\text{Câncer})}{P(\text{Positivo})}$$

$$P(\text{Câncer} | \text{Positivo}) = \frac{0.80 \times 0.01}{0.098} \approx 8,16\%$$

Surpreendente, não? Apesar do susto do teste positivo, a chance real de a pessoa ter a doença é de apenas **8,16%**! Isso acontece porque a doença original é rara (apenas 1% da população). É exatamente esse tipo de cálculo frio, preciso e baseado em evidências prévias que o Naïve Bayes usará para recomendar os nossos filmes e atuar em diagnósticos críticos.

## 3.3 Como a ciência avalia se o modelo é bom?
Quando estamos recomendando um filme da Netflix, recomendar o filme errado causa apenas um pequeno aborrecimento. No diagnóstico de um AVC, um erro pode ser fatal. Portanto, a ciência exige formas rigorosas de avaliar o classificador.

No artigo revisado, os pesquisadores notaram que a comunidade científica utiliza métricas específicas para validar as soluções. Não basta olhar apenas para a "porcentagem de acertos". Precisamos destrinchar a **Matriz de Confusão** e avaliar:

1. **Acurácia (Accuracy):** Das previsões totais (com AVC e sem AVC), quantas o modelo acertou?
2. **Precisão (Precision):** De todos os pacientes que o modelo disse "Tem AVC", quantos *realmente* tinham? (Mede o quanto o modelo dá falsos alarmes).
3. **Recall (Revocação / Sensibilidade):** De todos os pacientes que *realmente tinham* AVC, quantos o modelo conseguiu detectar? (Na medicina, essa costuma ser a métrica mais crítica. Um "Falso Negativo" significa mandar um paciente infartando para casa).
4. **F1-Score:** Uma média harmônica que tenta equilibrar a Precisão e o Recall, garantindo que o modelo é robusto em ambas as frentes.

Essas são as mesmas métricas que o `scikit-learn` gerou para nós quando testamos o reconhecimento de rostos com óculos no módulo de SVM, e serão as métricas que usaremos para avaliar o nosso recomendador de filmes!

# **Capítulo 4: Mão na Massa com Python (Construindo o nosso Recomendador)**  
Depois de explorarmos a matemática e vermos como o Naïve Bayes salva vidas na medicina, chegou a hora de aplicarmos este conhecimento. Vamos construir um sistema de recomendação de filmes utilizando um conjunto de dados real e muito famoso na comunidade de Data Science: o MovieLens 1M.  

Este conjunto possui cerca de 1 milhão de avaliações (de 1 a 5 estrelas) feitas por 6.040 utilizadores em 3.706 filmes. O nosso objetivo é simples: prever se um utilizador vai gostar de um determinado filme com base na forma como ele avaliou outros filmes. 

## 4.1 Preparando os Dados (O Ficheiro de Avaliações)
No nosso ficheiro descarregado, temos acesso ao ratings.dat, que contém as avaliações.  
Vamos usar o pandas e o numpy para carregar e explorar estes dados:

```python
import numpy as np
import pandas as pd

# Carregando os dados de avaliações
data_path = 'ml-1m/ratings.dat'
df = pd.read_csv(data_path, header=None, sep='::', engine='python')

# Nomeando as colunas para facilitar a leitura
df.columns = ['user_id', 'movie_id', 'rating', 'timestamp']
print(df.head())
```

Para termos a certeza do tamanho do nosso desafio, podemos verificar quantos utilizadores e filmes únicos existem:

```python
n_users = df['user_id'].nunique()
n_movies = df['movie_id'].nunique()

print(f"Número de utilizadores: {n_users}")
# Saída esperada: 6040
print(f"Número de filmes: {n_movies}")
# Saída esperada: 3706
```

## 4.2 Construindo a Matriz de Características (A Visão do Algoritmo)
O `scikit-learn` não entende tabelas soltas. Ele precisa de uma matriz matemática onde cada linha seja um utilizador e cada coluna seja um filme. O valor dentro da matriz será a nota que o utilizador deu ao filme (ou zero, se nunca o tiver visto).

```python
def load_user_rating_data(df, n_users, n_movies):
    # Cria uma matriz vazia (cheia de zeros)
    data = np.zeros([n_users, n_movies], dtype=np.intc)
    movie_id_mapping = {}
    
    # Preenche a matriz com as avaliações reais
    for user_id, movie_id, rating in zip(df['user_id'], df['movie_id'], df['rating']):
        user_id = int(user_id) - 1
        if movie_id not in movie_id_mapping:
            movie_id_mapping[movie_id] = len(movie_id_mapping)
        data[user_id, movie_id_mapping[movie_id]] = rating
        
    return data, movie_id_mapping

# Executando a função
data, movie_id_mapping = load_user_rating_data(df, n_users, n_movies)
```

# 4.3 Definindo o Alvo (O Filme "Cobaia")   
A maior parte da nossa matriz está cheia de zeros, pois é impossível que um utilizador tenha visto todos os 3.706 filmes. Para facilitar o nosso treino, vamos escolher o filme com o maior número de avaliações conhecidas para ser o nosso "Alvo" (Target).

```python
# Descobrindo qual filme tem mais avaliações
print(df['movie_id'].value_counts().head(5))
```

*O resultado mostrará que o filme com o ID 2858 tem o maior número de avaliações (3.428 avaliações).*  
Agora, separamos o filme `2858` como o nosso alvo ``(Y)`` e as avaliações de todos os outros filmes como as características preditivas ``(X)``:   

```python
target_movie_id = 2858

# X: Todos os filmes, exceto o filme alvo
X_raw = np.delete(data, movie_id_mapping[target_movie_id], axis=1)

# Y: Apenas as avaliações do filme alvo
Y_raw = data[:, movie_id_mapping[target_movie_id]]

# Filtramos apenas os utilizadores que realmente avaliaram o filme alvo
X = X_raw[Y_raw > 0]
Y = Y_raw[Y_raw > 0]
```

### A Magia da Classificação Binária  

Lembra-se da Classificação Binária que vimos no Capítulo 1? Vamos transformar as notas de 1 a 5 num problema simples de Gostou (1) ou Não Gostou (0). Consideramos que notas acima de 3 indicam que o filme foi "curtido".   

```python
recommend = 3
Y[Y <= recommend] = 0
Y[Y > recommend] = 1

n_pos = (Y == 1).sum()
n_neg = (Y == 0).sum()
print(f'{n_pos} amostras positivas e {n_neg} amostras negativas.')
# Saída esperada: 2853 amostras positivas e 575 amostras negativas.
```

## 4.4 O Perigo do Desbalanceamento de Dados  
Reparem no resultado acima: temos 2.853 pessoas que gostaram do filme e apenas 575 que não gostaram. O nosso conjunto de dados está desbalanceado!   
Se não tivermos cuidado, o algoritmo pode simplesmente "viciar-se" e prever sempre que toda a gente gosta do filme, ignorando a minoria. No mundo real (e na saúde, como vimos com o AVC), isso é um erro fatal. Algumas estratégias para corrigir isto:   
* Oversampling: Duplicar instâncias da classe minoritária ou gerar dados sintéticos.   
* Undersampling: Remover aleatoriamente amostras da classe maioritária para igualar os volumes.   
* Class Weighting: Atribuir "pesos" matemáticos maiores aos erros cometidos na classe minoritária durante o treino.   
## 4.5 Separando o Treino e o Teste (O Simulado Final)  
Para avaliar de forma justa o nosso classificador, devemos esconder uma parte dos dados (Conjunto de Teste). Para conjuntos de dados pequenos, separar 25% a 30% para teste é o ideal. Para conjuntos grandes, 20% é suficiente.  
Vamos usar o famoso ``train_test_split`` do ``scikit-learn``:   

```python
from sklearn.model_selection import train_test_split

# Separando 80% para treino e 20% para teste
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print(f"Treino: {len(Y_train)} | Teste: {len(Y_test)}")
```
